```python
import numpy as np
import matplotlib.pyplot as plt

# x values
x = np.linspace(-5, 5, 200)

# Equations:
# 2x + 3y = 3   => y = (3 - 2x)/3
# 4x + 6y = 6   => y = (6 - 4x)/6
# 3x - 2y = 1   => y = (3x - 1)/2

y1 = (3 - 2*x) / 3
y2 = (6 - 4*x) / 6
y3 = (3*x - 1) / 2

# Plot lines
plt.plot(x, y1, label="2x + 3y = 3")
plt.plot(x, y2, label="4x + 6y = 6")
plt.plot(x, y3, label="3x - 2y = 1")

# Intersection of first two lines
# Since second equation is multiple of first,
# both overlap and intersect with third at (1, 1/3)

plt.scatter(1, 1/3, color="red")
plt.text(1, 1/3, " (1, 1/3)")

# Axes
plt.axhline(0, color="black")
plt.axvline(0, color="black")

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.grid(True)
plt.legend()
plt.title("Graphical Solution of Linear System")
plt.xlabel("x")
plt.ylabel("y")

plt.show()

print("Solution Point = (1, 1/3)")
```

```python
import numpy as np

# Create random 5x5 matrix
A = np.random.randint(1, 10, (5, 5))

print("Matrix A:\n", A)

# Rank
rank = np.linalg.matrix_rank(A)
print("\nRank of A:", rank)

# Determinant
det = np.linalg.det(A)
print("Determinant of A:", det)

# Invertibility check
if det != 0:
    print("Matrix is invertible")
    A_inv = np.linalg.inv(A)
    print("Inverse of A:\n", A_inv)
else:
    print("Matrix is NOT invertible")
```

```python
import numpy as np

# Random 5x5 matrix
A = np.random.randint(1, 10, (5, 5))

# Random vector of size 5
x = np.random.randint(1, 10, 5)

print("Matrix A:\n", A)
print("\nVector x:\n", x)

# Compute b = Ax
b = np.dot(A, x)
print("\nVector b = Ax:\n", b)

# Solve Ax = b
try:
    x_solution = np.linalg.solve(A, b)
    print("\nSolution of Ax = b:\n", x_solution)
except np.linalg.LinAlgError:
    print("\nMatrix is singular, cannot solve uniquely")
```

```python
import numpy as np

# Coefficient matrix
A = np.array([
    [8, 3, -2],
    [-4, 7, 5],
    [3, 4, -12]
])

# Constants vector
b = np.array([9, 15, 35])

# Solve system
solution = np.linalg.solve(A, b)

print("Solution (x, y, z):\n", solution)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Given data points
x = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])

# Step 1: Plot the points

plt.scatter(x, y, color='black', label='Data Points')
plt.title("Given Points")
plt.xlabel("X")
plt.ylabel("Y")


# Step 2: Guess polynomial
# Since 10 points → max degree = 9

degree = 3  # Try cubic first (simpler guess)


# Step 3: Solve using linear system (Vandermonde matrix)

A = np.vander(x, degree + 1)  # Creates matrix for polynomial
coeffs = np.linalg.lstsq(A, y, rcond=None)[0]

print("Polynomial Coefficients (highest power first):")
print(coeffs)

# Step 4: Generate fitted curve

x_fit = np.linspace(min(x), max(x), 200)
y_fit = np.polyval(coeffs, x_fit)

plt.plot(x_fit, y_fit, label=f"Fitted Polynomial (degree {degree})")

# Step 5: Check fit accuracy

y_pred = np.polyval(coeffs, x)
error = np.linalg.norm(y - y_pred)

print("\nFitting Error:", error)

plt.legend()
plt.grid()
plt.show()
```
